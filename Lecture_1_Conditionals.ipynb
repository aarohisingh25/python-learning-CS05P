MATCH - just like switch

match variable:
  case "1":
    statement
  case "2":
    statement

```python
name=input("Enter your name=")

match name:
  case "Harry":
    print("Gryffindor")
  case "Hermione":
    print("Gryffindor")
  case "Ron":
    print("Gryffindor")
  case "Draco":
    print("Slytherin")
  case _:
    print("Who?")
```

```python
name=input("Enter your name=")

match name:
  case "Harry" | "Hermione" | "Ron":
    print("Gryffindor")
  case "Draco":
    print("Slytherin")
  case _:
    print("Who?")
```
