Write code with ErrorHandling

```python
try:
  x=int(input("What's x?"))
  print(f"x is {x}")
except ValueError:
  print("x is not an integer")
```

If the user enters hello, Python cannot convert it into an integer, so a ValueError occurs.
But x was never successfully assigned a value, so trying to print {x} inside except can cause a NameError

```python
try:
  x=int(input("What's x?"))
  print(f"x is {x}")
except ValueError:
  print(f"{x} is not an integer")
```

```python
try:
  x=int(input("What's x?"))
except ValueError:
  print("x is not an integer")

print(f"x is {x}")
```


try: Executes code that might cause an error.

except: Handles the error if one occurs.

else: Executes only if no error occurs in the try block.

```python
try:
  x=int(input("What's x?"))
except ValueError:
  print("x is not an integer")
else:
  print(f"x is {x}")
```

```python
while True:
  try:
    x=int(input("What's x?"))
  except ValueError:
    print("x is not an integer")
  else:
    break

print(f"x is {x}")
```

```python
while True:
  try:
    x=int(input("What's x?"))
    break
  except ValueError:
    print("x is not an integer")

print(f"x is {x}")
```

```python
def main():
  x=get_int()
  print(f"x is {x}")

def get_int():
  while True:
    try:
      x=int(input("What's x?"))
    except ValueError:
      print("x is not an integer")
    else:
      return x

main()
```

pass -pass is a placeholder statement in Python that does nothing when executed, allowing you to leave a block empty without causing an error.

```python
def main():
  x=get_int()
  print(f"x is {x}")

def get_int():
  while True:
    try:
      return int(input("What's x?"))
    except ValueError:
      pass

main()
```

Function Arguements

```python
def main():
  x=get_int("What's x?")
  print(f"x is {x}")

def get_int(prompt):
  while True:
    try:
      return int(input(prompt))
    except ValueError:
      pass

main()
```
