STRING METHODS

1.string strip-remove unnecessary spaces from the starting

here we have lstrip() and rstrip() depending on how we want to remove the spaces

```python
name=input("Enter your full name:-")
name=name.strip()
print("Hello",name)
```

```python
name=input("Enter your full name:-")
name=name.lstrip()
print("Hello",name)
```

```python
name=input("Enter your full name:-")
name=name.rstrip()
print("Hello",name)
```

2.string capitalize-make the first letter of the first word capital

```python
name=input("Enter your full name:-")
name=name.capitalize()
print("Hello",name)
```

3.string title-make the first letter of every word capital

```python
x=input("Enter a quote:-")
x=x.title()
print(x)
```

```python
name=input("Enter your name =").strip().title()
print(f"Hello,{name}")
```

4.string split-split string into multiple substrings

```python
name=input("Enter your full name:-")
first,last=name.split(" ")
print(f"\nHello,{first}")
print(f"Your last name is {last}")
```
