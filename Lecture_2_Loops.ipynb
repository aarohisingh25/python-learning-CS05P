
break: Terminates the loop immediately.

continue: Skips the current iteration and moves to the next iteration.

```python
while True:
  n=int(input("What's n? "))
  if n<0:
    continue
  else:
    break

for i in range(n):
  print("meow")
```

```python
def main():
  print_square(3)

def print_square(size):

  #for each row in square
  for i in range("size"):

    #for each brick in row
    for j in range("size"):

      #print brick
      print("#",end="")

    print()

main()
```
