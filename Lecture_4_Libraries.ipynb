slice of a list means to take subset of a list

1.[start:end]
2.[start:]

positive slicing from 0 to n
negative slicing from -(n+1) to -1

example PYTHON

positive index:-
P=0
Y=1
T=2
H=3
O=4
N=5

negative index:-
P=-6
Y=-5
T=-4
H=-3
O=-2
N=-1


Code	Output	Meaning
s[-3:]	HON	Last 3 characters
s[:-2]	PYTH	Everything except last 2
s[-4:-1]	THO	From index -4 to before -1
s[::-1]	NOHTYP	Reverse the string
s[-1:]	N	Last character as a string
s[:-1]	PYTHO	Remove last character




```python
import sys

if len(sys.argv)<2:
  sys.exit("Too less arguement")

for arg in sys.argv[1:]:
  print("Hello my name is",arg)
```
